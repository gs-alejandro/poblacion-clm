# 4. Pipelines

```python
from sklearn.pipeline import Pipeline
from sklearn.decomposition import PCA
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, FunctionTransformer, PowerTransformer
```

## 4.1. PCA

```python
log_cols = [
    "total", 
    "densidad_pob", 
    "p_cambio_total_10", 
    # "p_cambio_total_10_prev", 
    "p_cambio_total_20",
    # "p_saldo_ext_10", "p_saldo_int_10",
]
# con transformación logarítmica los casos extremos seguían dominando la dispersión en PC2, causando clusters de 1 municipio
saldo_cols = ["p_saldo_ext_10", "p_saldo_int_10"] 

pre = ColumnTransformer(
    [
        ("log", FunctionTransformer(np.log1p, feature_names_out="one-to-one"), log_cols),
        ("yj", PowerTransformer(method="yeo-johnson"), saldo_cols),
    ],
    remainder="passthrough",
    verbose_feature_names_out=False
)
pca_pipe = Pipeline([
    ("pre", pre),
    ("scaler", StandardScaler()),
    ("pca", PCA(n_components=3, random_state=42))
]).set_output(transform="pandas")
```

# 5. PCA

## 5.1. Método del codo

```python
from sklearn.base import clone
```

```python
pca_pipe_full = clone(pca_pipe).set_params(pca__n_components=None)
pca_pipe_full.fit(c_df)
pca_full = pca_pipe_full.named_steps["pca"]

var = pca_full.explained_variance_ratio_
cum = var.cumsum()
n = np.arange(1, len(var) + 1)

fig, ax1 = plt.subplots(figsize=(11, 5))

ax1.bar(n, var, alpha=0.5, label="Varianza individual")
ax1.plot(n, var, "o-", color="C0")
ax1.set_xlabel("Nº de componentes")
ax1.set_ylabel("Varianza explicada")
ax1.set_xticks(n)

ax2 = ax1.twinx()
ax2.plot(n, cum, "s-", color="C1", label="Varianza acumulada")
ax2.set_ylabel("Varianza acumulada")
# ax2.set_ylim(0, 1.05)

h1, l1 = ax1.get_legend_handles_labels()
h2, l2 = ax2.get_legend_handles_labels()
ax1.legend(h1 + h2, l1 + l2, loc="center right")

for x, v, c in zip(n, var, cum):
    # Desplazamiento especial para x=1 para evitar que los textos se pisen
    offset_var = (-12, 6) if x == 1 else (0, 6)
    offset_cum = (12, 6) if x == 1 else (0, 6)

    # Etiqueta varianza individual
    ax1.annotate(
        f"{v:.2f}",
        (x, v),
        textcoords="offset points",
        xytext=offset_var,
        ha="center",
        fontsize=8.5,
        color="C0",
        weight="bold",
    )

    # Etiqueta varianza acumulada
    ax2.annotate(
        f"{c:.2f}",
        (x, c),
        textcoords="offset points",
        xytext=offset_cum,
        ha="center",
        fontsize=8.5,
        color="C1",
        weight="bold",
    )

plt.title("Varianza explicada por nº de componentes")
plt.tight_layout()
plt.show()
```

# 6. Clustering

```python
from sklearn.cluster import KMeans, AgglomerativeClustering, HDBSCAN
from sklearn.metrics import silhouette_score
```

## 6.1 KMeans

```python
kmeans_models = []
kmeans_inertias = []
kmeans_sils = []
kmeans_dbs = []

for k in range(2, 26):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X)
    kmeans_models.append(kmeans)
    kmeans_inertias.append(kmeans.inertia_)
    kmeans_sils.append(silhouette_score(X, kmeans.labels_))
```

```python
fig, ax = plt.subplots(figsize=(11, 5))

ax.bar(range(2, 26), kmeans_inertias, alpha=0.5)
ax.plot(range(2, 26), kmeans_inertias, "o-", color="C0")
ax.set_xlabel("Nº de grupos")
ax.set_ylabel("Inercia")
ax.set_xticks(range(2, 26))

plt.title("Inercia - KMeans")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(11, 5))

ax.bar(range(2, 26), kmeans_sils, alpha=0.5)
ax.plot(range(2, 26), kmeans_sils, "o-", color="C0")
ax.set_xlabel("Nº de grupos")
ax.set_ylabel("Silhouette")
ax.set_xticks(range(2, 26))

plt.title("Silhouette - KMeans")
plt.show()
```

## 6.2. Agglomerative clustering (jerárquico)

```python
agglo_models = []
agglo_inertias = []
agglo_sils = []
agglo_dbs = []

for k in range(2, 26):
    agglo = AgglomerativeClustering(n_clusters=k, linkage="ward", metric="euclidean")
    agglo.fit(X)
    agglo_models.append(agglo)
    # agglo_inertias.append(agglo.inertia_)
    agglo_sils.append(silhouette_score(X, agglo.labels_))
```

```python
fig, ax = plt.subplots(figsize=(11, 5))

ax.bar(range(2, 26), agglo_sils, alpha=0.5)
ax.plot(range(2, 26), agglo_sils, "o-", color="C0")
ax.set_xlabel("Nº de grupos")
ax.set_ylabel("Silhouette")
ax.set_xticks(range(2, 26))

plt.title("Silhouette - Agglomerative clustering")
plt.show()
```

```python

K_KMEANS = 4   # nº de grupos elegido para KMeans
K_AGGLO = 4    # nº de grupos elegido para el jerárquico

resultados = {
    f"KMeans (k={K_KMEANS})": kmeans_models[K_KMEANS - 2].labels_,
    f"Jerárquico Ward (k={K_AGGLO})": agglo_models[K_AGGLO - 2].labels_,
    "HDBSCAN": hdbscan.labels_,
}


def plot_clusters_3d(X, labels, titulo):
    df_plot = X.iloc[:, :3].copy()
    df_plot.columns = ["PC1", "PC2", "PC3"]
    df_plot["Cluster"] = np.where(labels == -1, "Ruido", labels.astype(str))

    sil = silhouette_score(X.iloc[:, :3], labels)

    fig = px.scatter_3d(
        df_plot.reset_index(),
        x="PC1", y="PC2", z="PC3",
        color="Cluster",
        hover_name="cod_mun",
        opacity=0.7,
        title=f"{titulo} - Silhouette: {sil:.3f}",
    )
    fig.update_traces(marker_size=3)
    fig.update_layout(height=650, legend_itemsizing="constant",
                      margin=dict(l=0, r=0, t=50, b=0))
    fig.show()
```

```python
for titulo, labels_metodo in resultados.items():
    plot_clusters_3d(X, labels_metodo, titulo)
```

```python
df = pd.read_csv("data/pob_clm_clusters.csv", dtype={"cod_mun": str, "cod_prov": str})
df = df[(df["periodo"] == 2025) & df["cluster"].notna()].copy()

df["p_saldo_ext_10"] = df["p_suma_altas_ext_10"] - df["p_suma_bajas_ext_10"]
df["p_saldo_int_10"] = df["p_suma_altas_int_10"] - df["p_suma_bajas_int_10"]

medianas_reales = df.groupby("cluster")[view_cols].median()
df_norm = pd.DataFrame(
    {c: [percentileofscore(df[c].dropna(), v, kind="mean") / 100 for v in medianas_reales[c]]
     for c in view_cols},
    index=medianas_reales.index,
)
df_norm
```

```python
col_cluster = "cluster"

variables = list(df_norm.columns)

etiquetas = {
    "total": "Población total",
    "densidad_pob": "Densidad de población",
    "p_cambio_total_10": "Prop. cambio en 10 años",
    "p_cambio_total_20": "Prop. cambio en 20 años",
    "p_ed_65+": "Prop. 65 años o más",
    "p_ed_16_64": "Prop. 16 a 64 años",
    "p_ed_0_15": "Prop. 0 a 15 años",
    "p_saldo_ext_10": "Prop. saldo migratorio exterior 10 años",
    "p_saldo_int_10": "Prop. saldo migratorio interior 10 años",
    "p_suma_mnp_crecimiento_veg_10": "Prop. crecimiento vegetativo 10 años",
}
nombres = [etiquetas.get(v, v) for v in variables]
theta_cerrado = nombres + [nombres[0]]
paleta = ["#0072B2", "#E69F00", "#009E73", "#CC79A7", "#D55E00"]
estilos = ["solid", "dash", "dot", "dashdot", "longdash"]

medianas_reales = (
    df.groupby(col_cluster)[variables]
      .median()
      .reindex(df_norm.index)
)
n_por_cluster = df[col_cluster].value_counts()

fig = go.Figure()

for i, cluster in enumerate(df_norm.index):
    r = df_norm.loc[cluster, variables].to_numpy()
    medias = medianas_reales.loc[cluster, variables].to_numpy()
    color = paleta[i % len(paleta)]

    detalle = "<br>".join(
        f"  {nom}: {val:,.2f}" for nom, val in zip(nombres, medias)
    )
    nombre = f"<b>Cluster {int(cluster)}</b> (n={n_por_cluster.get(cluster, 0)})<br>{detalle}"

    fig.add_trace(go.Scatterpolar(
        r=np.append(r, r[0]),
        theta=theta_cerrado,
        customdata=np.append(medias, medias[0]),
        hovertemplate=(
            f"<b>Cluster {int(cluster)}</b><br>"
            "%{theta}<br>"
            "Mediana real: %{customdata:,.2f}<br>"
            "Percentil: %{r:.0%}"
            "<extra></extra>"
        ),
        fill=None,
        name=nombre,
        line=dict(
            color=color, width=3,
            # dash=estilos[i % len(estilos)]
        ),
        opacity=1,
        legendgroup=i
    ))

fig.update_layout(
    title=dict(text="Percentil de mediana por clúster", x=0.5, font=dict(size=20)),
    width=1800,
    height=1200,
    polar=dict(radialaxis=dict(
        visible=False, range=[0, 1]
    )),
    showlegend=True,
    legend=dict(
        title="", font=dict(size=14),
        tracegroupgap=20
    ),
)

fig.show()
```

```python
import geopandas as gpd
from matplotlib.patches import Patch

NOMBRES_TIPOS = {"A": "Rural receptor", "B": "Rural en declive", "C": "Intermedio arraigado", "D": "En expansión"}
COLORES_TIPOS = ["#0072B2", "#E69F00", "#009E73", "#CC79A7"]

# Pair cluster indices (0, 1, 2, 3) with color codes
color_map = dict(enumerate(COLORES_TIPOS))
letras = list(NOMBRES_TIPOS.keys())

tipos = (
    pd.read_csv("data/pob_clm_clusters.csv", dtype={"cod_mun": str}, usecols=["cod_mun", "cluster"])
    .dropna()
    .drop_duplicates("cod_mun")
    .set_index("cod_mun")["cluster"]
    .astype(int)
)

mapa_mun = gpd.read_parquet("data/region_municipios_clm.parquet")
mapa_prov = gpd.read_parquet("data/region_provincias_clm.parquet")

mapa_mun["color"] = mapa_mun["cod_mun"].map(tipos).map(color_map).fillna("#DDDDDD")
print("Municipios sin tipo:", mapa_mun["cod_mun"].map(tipos).isna().sum())

fig, ax = plt.subplots(figsize=(12, 8))
mapa_mun.plot(ax=ax, color=mapa_mun["color"], edgecolor="white", linewidth=0.1)
mapa_prov.boundary.plot(ax=ax, color="black", linewidth=0.6)

# Build legend handles by matching key letters ('A', 'B', ...) with colors
handles = [
    Patch(color=c, label=f"Grupo {letra}: {NOMBRES_TIPOS[letra]}")
    for letra, c in zip(letras, COLORES_TIPOS)
]
ax.legend(handles=handles, loc="lower left", frameon=False)

ax.set_axis_off()
fig.tight_layout()

fig.savefig("figs/mapa_tipos.pdf", bbox_inches="tight")
plt.show()
```

```python
d = pd.read_csv("data/pob_clm_clusters.csv", dtype={"cod_mun": str, "cod_prov": str})
d = d[(d["periodo"] == 2025) & d["cluster"].notna()].copy()
d["cluster"] = d["cluster"].astype(int)
d["p_saldo_ext_10"] = d["p_suma_altas_ext_10"] - d["p_suma_bajas_ext_10"]
d["p_saldo_int_10"] = d["p_suma_altas_int_10"] - d["p_suma_bajas_int_10"]

FILAS = [  # (etiqueta, columna, decimales, en porcentaje)
    ("Población (habitantes)", "total", 0, False),
    ("Densidad (hab./km$^2$)", "densidad_pob", 1, False),
    ("Nacidos en el municipio", "p_n_en_mun", 1, True),
    ("65 o más años", "p_ed_65+", 1, True),
    ("0 a 15 años", "p_ed_0_15", 1, True),
    ("Cambio en 10 años", "p_cambio_total_10", 1, True),
    ("Cambio en 20 años", "p_cambio_total_20", 1, True),
    ("Saldo exterior (10 años)", "p_saldo_ext_10", 1, True),
    ("Saldo interior (10 años)", "p_saldo_int_10", 1, True),
    ("Crecimiento vegetativo (10 años)", "p_suma_mnp_crecimiento_veg_10", 1, True),
]

def fmt(v, dec, pct):
    """Coma decimal, punto de miles y signo menos de LaTeX."""
    if pd.isna(v):
        return "--"
    v = v * 100 if pct else v
    s = f"{abs(v):,.{dec}f}".replace(",", "X").replace(".", ",").replace("X", ".")
    return ("$-$" if v < 0 else "") + s + (r"\,\%" if pct else "")

cols = [c for _, c, _, _ in FILAS]
med = pd.concat([d.groupby("cluster")[cols].median(), d[cols].median().to_frame("Total").T])
n = pd.concat([d["cluster"].value_counts().sort_index(), pd.Series({"Total": len(d)})])
display(med)

tipos = [k for k in med.index if k != "Total"]
lineas = [
    r"\begin{table}[h]\footnotesize",
    r"\renewcommand{\arraystretch}{1.3}",
    r"\begin{center}",
    r"\begin{tabular}{|l|" + "r|" * len(med) + "}",
    r"\hline",
    r"\bft{Variable} & " + " & ".join(rf"\bft{{Tipo {k}}}" for k in tipos) + r" & \bft{Total} \\ \hline",
    "Municipios & " + " & ".join(str(n[k]) for k in med.index) + r" \\ \hline",
]
for etiqueta, c, dec, pct in FILAS:
    lineas.append(f"{etiqueta} & " + " & ".join(fmt(med.at[k, c], dec, pct) for k in med.index) + r" \\ \hline")
lineas += [
    r"\end{tabular}",
    r"\end{center}",
    r"\caption[Perfil de los tipos de municipio]{Mediana de cada variable en 2025 por tipo de municipio.}",
    r"\label{tab:perfiles}",
    r"\renewcommand{\arraystretch}{1}",
    r"\end{table}",
]
print("\n".join(lineas))
```
