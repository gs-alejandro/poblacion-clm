## 3.1. Descripción por percentiles

## 3.2. Gráfico de densidad 

### 3.2.1. Total

```python
fig, ax1 = plt.subplots(figsize=(10, 5))
sns.kdeplot(a.loc[a["periodo"] == 2025, "p_cambio_total_20"], fill=True)
ax1.axvline(a.loc[a["periodo"] == 2025, "p_cambio_total_20"].median(), linestyle="--", alpha=0.9, label="Mediana")
ax1.axvline(a.loc[a["periodo"] == 2025, "p_cambio_total_20"].mean(), linestyle="-.", alpha=0.9, color="orange", label="Media")
ax1.legend()
ax1.set_ylabel("Densidad")
ax1.set_ylim(-0.01, None)
plt.show()
```

```python
fig, ax1 = plt.subplots(figsize=(10, 5))
# sns.kdeplot(a["p_cambio_total_20"], fill=True, ax=ax1); ax1.set_xlim(-2, 3)
sns.kdeplot(a.loc[(a["periodo"] == 2025) & (a["p_cambio_total_20"] < 3), "p_cambio_total_20"], fill=True, ax=ax1)
ax1.axvline(a.loc[a["periodo"] == 2025, "p_cambio_total_20"].median(), linestyle="--", alpha=0.9, label="Mediana")
ax1.axvline(a.loc[a["periodo"] == 2025, "p_cambio_total_20"].mean(), linestyle="-.", alpha=0.9, color="orange", label="Media")
ax1.set_ylabel("Densidad")
ax1.legend()
plt.show()
```

# 4. Correlaciones con el crecimiento

```python
cols = [
    "p_cambio_total_20", "p_cambio_total_10",
    "p_n_ext", "p_n_en_mun",
    "p_ed_0_15", "p_ed_16_64", "p_ed_65+",
    # "p_saldo_int_10", "p_saldo_ext_10",
    "p_ae_agricultura_ganaderia_pesca",
]

fig, ax1 = plt.subplots(figsize=(10, 8))

sns.heatmap(
    a.loc[a["periodo"] == 2025, cols].corr(),
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    annot=True,
    fmt=".2f",
    ax=ax1,
)

plt.tight_layout()
plt.show()
```
