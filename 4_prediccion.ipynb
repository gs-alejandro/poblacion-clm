# 5. Modelo Holt

```python
from pathlib import Path
from statsmodels.tsa.holtwinters import Holt
```

```python
PROV_CLM = ("02", "13", "16", "19", "45")
ANIO_FIN = int(df["periodo"].max())
```

```python
def fit_holt(y, phi=None):
    return Holt(y, damped_trend=True, initialization_method="estimated").fit(damping_trend=phi)

def predict_log(ts_log, phi=None, H=10):
    anios = ts_log.columns.max() + np.arange(1, H + 1)
    preds, params = {}, {}
    for cod, y in ts_log.iterrows():
        m = fit_holt(y.dropna().to_numpy(), phi)
        preds[cod] = m.forecast(H)
        params[cod] = {"alpha": m.params["smoothing_level"],
                       "beta": m.params["smoothing_trend"],
                       "phi": m.params["damping_trend"]}
    preds = pd.DataFrame(preds, index=anios).T.rename_axis("cod_mun")
    params = pd.DataFrame(params).T.rename_axis("cod_mun")
    return preds, params
```

# 6. Elección de configuración

```python
ORIGENES = range(2011, 2021)
H = 10
CONFIGS = {"holt": None, "holt_phi": 0.9}
```

```python
MODELO = "holt" 
# MODELO = "holt_phi" 
PHI = CONFIGS[MODELO]
```

# 7. Modelo final

## 7.1. Ajuste completo

```python
pred_log, params_final = predict_log(ts_log, PHI, H=10)
params_final.describe()
```

## 7.2. Tabla de predicciones a 10 años

```python
ANIOS_PRED = range(ANIO_FIN+1, ANIO_FIN + 11)
col_base = f"pob_{ANIO_FIN}"

pred = (df.loc[df["periodo"] == ANIO_FIN, ["cod_mun", "nombre", "cod_prov", "total"]]
        .rename(columns={"total": col_base})
        .set_index("cod_mun")
        .loc[pred_log.index])

for anio in ANIOS_PRED:
    h = anio - ANIO_FIN
    p = pred_log[anio]
    pred[f"pob_{anio}"] = np.exp(p)
    pred[f"tasa_anual_{anio}"] = (pred[f"pob_{anio}"] / pred[col_base]) ** (1 / h) - 1

pred = pred.reset_index()
pred.head()
```
